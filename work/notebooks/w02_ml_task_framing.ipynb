## Unit of Analysis

Each row in the dataset represents **one content page**.

Each page contains search performance, engagement, content quality, and metadata features that can be used to estimate its refresh priority.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content_refresh_anonymized.csv")

# Show dataset information
print("Dataset Shape:", df.shape)

# Create a target column from trend_direction
df["target"] = (df["trend_direction"] == "down").astype(int)

# Show target distribution
print("\nTarget Distribution:")
print(df["target"].value_counts())

# Display one row = one content page
display(df[["content_id", "trend_direction", "target"]].head())
```

## Why ML Beats a Fixed Rule

A simple rule such as "refresh pages older than 180 days" ignores many important signals.

Page performance depends on multiple factors including impressions, CTR, average position, search volume, content age, engagement, and user behavior.

Machine learning can combine these signals and rank pages more accurately than a single manually written rule. The output supports SEO managers by providing a prioritized review queue instead of relying on one fixed condition.
