# Ranked actions + reason codes

This playbook converts the validated model output into actionable SEO recommendations.

| Rank | Action | Reason Code |
|------|---------|-------------|
| High | Refresh content | REFRESH_HIGH_VOLUME_LOW_IMPRESSIONS |
| High | Improve title/meta description | LOW_CTR |
| Medium | Update outdated information | STALE_CONTENT |
| Medium | Expand keyword coverage | LOW_VISIBILITY |
| Low | Monitor performance | STABLE_PAGE |

### Archetype → Action

- High search volume + low impressions → Refresh content
- Low CTR → Improve title and meta description
- Older content → Update content freshness
- Stable pages → Continue monitoring

### Decay / Refresh Insight

Pages that combine high search demand with declining visibility are stronger refresh candidates than pages with consistently low demand.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import os

df = pd.read_csv("/content/content_refresh_anonymized.csv")

df["baseline_score"] = (
    df["search_volume"].rank(pct=True) * 0.5 +
    (1 - df["ctr"].rank(pct=True)) * 0.3 +
    (1 - df["impressions_90d"].rank(pct=True)) * 0.2
)

df["reason_code"] = "REFRESH_HIGH_VOLUME_LOW_IMPRESSIONS"
df["action"] = "Refresh Content"

queue = df.sort_values(
    "baseline_score",
    ascending=False
)

display(queue.head(10))
```
